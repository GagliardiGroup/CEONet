## Reproduce Training Curve

```python
import matplotlib.pyplot as plt
import numpy as np
import torch
from deeporb.default_net import get_model, get_data
from deeporb.model_params import get_params

for d1 in ["sto3g","qh9","tm"]:
    for d2 in ["occ","virt"]:
        data = get_data(d1,d2)
        for t in [1000,10000,100000]:
            model = get_model(d1,d2,t)
            params = get_params(d1,d2)
            avge0, sigma = params["AVGE0"], params["SIGMA"]
            model.cuda()
            trues, preds = [], []
            for batch in data.val_dataloader():
                batch.cuda()
                out = model(batch)
                true = out["pred_energy"]*sigma + avge0
                pred = batch["energy"]
                trues.append(true.detach().cpu().numpy())
                preds.append(pred.detach().cpu().numpy())
                torch.cuda.empty_cache()
            trues = np.hstack(trues)
            preds = np.hstack(preds)
            n = int(t / 1000)
            np.save(f"../model_eval/orbital_energy_predictions/{n}k/{d1}_{d2}_val_true.npy",trues)
            np.save(f"../model_eval/orbital_energy_predictions/{n}k/{d1}_{d2}_val_pred.npy",preds)
            print(d1,d2,"val",np.abs(preds - trues).mean())
    
            trues, preds = [], []
            for batch in data.test_dataloader():
                batch.cuda()
                out = model(batch)
                true = out["pred_energy"]*sigma + avge0
                pred = batch["energy"]
                trues.append(true.detach().cpu().numpy())
                preds.append(pred.detach().cpu().numpy())
                torch.cuda.empty_cache()
            trues = np.hstack(trues)
            preds = np.hstack(preds)
            np.save(f"../model_eval/orbital_energy_predictions/{n}k/{d1}_{d2}_test_true.npy",trues)
            np.save(f"../model_eval/orbital_energy_predictions/{n}k/{d1}_{d2}_test_pred.npy",preds)
            print(d1,d2,"test",np.abs(preds - trues).mean())
                
```
